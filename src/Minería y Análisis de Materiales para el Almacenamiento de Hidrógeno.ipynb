# **1. Introducción: El Problema del Almacenamiento de Hidrógeno.**

## **¿Puede un automóvil funcionar con agua?**

Imaginemos un automóvil que, en lugar de quemar gasolina dentro de un motor, utiliza hidrógeno para producir electricidad y alimentar un motor eléctrico. Por el tubo de escape no expulsaría dióxido de carbono procedente de la reacción electroquímica, sino principalmente agua. La idea parece sencilla: almacenar hidrógeno en el vehículo, convertir su energía química en electricidad y utilizarla para desplazarse.

Sin embargo, esta imagen plantea una primera pregunta: **si el hidrógeno puede utilizarse para producir electricidad de manera eficiente, ¿por qué no se ha convertido en el combustible dominante del transporte?**

La respuesta comienza con una distinción fundamental: el hidrógeno no es una fuente primaria de energía en el mismo sentido que la energía solar o la energía eólica. En la mayoría de sus aplicaciones actuales, funciona como un **vector energético**: permite almacenar y transportar energía que tuvo que suministrarse previamente para producirlo. Por ejemplo, es posible obtenerlo mediante la electrólisis del agua, un proceso que utiliza electricidad para separar sus componentes. La reacción global puede escribirse como:


$$2H_2O(l)\rightarrow 2H_2(g)+O_2(g)$$

Aunque esta ecuación parece simple, separar el agua requiere aportar energía para superar la estabilidad de sus enlaces químicos. Por tanto, el hidrógeno no aparece gratuitamente al llenar un depósito: primero hay que producirlo, después acondicionarlo y almacenarlo, y finalmente recuperar parte de la energía invertida.

Cuando el hidrógeno llega a una celda de combustible, ocurre el proceso inverso en términos globales. El hidrógeno reacciona electroquímicamente con el oxígeno del aire y la celda transforma parte de la energía química liberada en electricidad. En una celda de combustible de membrana de intercambio protónico, conocida como PEMFC, las reacciones principales son:

En el ánodo, el hidrógeno se oxida:

$$2H_2(g)\rightarrow 4H^++4e^-$$

En el cátodo, el oxígeno se reduce:

$$O_2(g)+4H^++4e^-\rightarrow 2H_2O(l)$$

Los electrones no atraviesan directamente la membrana: circulan por un circuito externo y pueden alimentar el motor eléctrico del automóvil. Mientras tanto, los protones atraviesan la membrana y participan en la formación de agua en el cátodo.

La reacción global es:

$$2H_2(g)+O_2(g)\rightarrow 2H_2O(l)+\text{energía}$$

Esta transformación permite generar electricidad sin recurrir a la combustión convencional del combustible dentro de un motor. No obstante, que el vehículo no emita dióxido de carbono por el escape durante su funcionamiento no significa que todo el proceso esté libre de emisiones. El resultado ambiental depende, entre otros factores, de cómo se produjo el hidrógeno, de la electricidad utilizada y de la fabricación de los equipos.

El principio electroquímico, entonces, es prometedor. Pero todavía no hemos resuelto el problema que aparece antes de encender el automóvil.

## **El verdadero desafío: llevar suficiente hidrógeno a bordo**

El hidrógeno es el elemento más ligero de la tabla periódica. Esta característica le permite ofrecer una elevada cantidad de energía por unidad de masa. Sin embargo, su densidad en condiciones ambientales es extremadamente baja: un kilogramo de hidrógeno ocupa un volumen considerable si se encuentra como gas a presión cercana a la atmosférica.

Para un automóvil, esto representa un problema práctico. No basta con que un combustible almacene mucha energía por kilogramo; también debe ser posible transportar suficiente combustible en un espacio limitado, sin añadir una masa excesiva ni comprometer la seguridad.

Una alternativa consiste en comprimir el hidrógeno en recipientes especialmente diseñados para soportar altas presiones. Otra es enfriarlo hasta temperaturas criogénicas para almacenarlo como líquido. Ambas estrategias permiten aumentar la cantidad de hidrógeno contenida en un volumen determinado, pero introducen dificultades adicionales: los recipientes presurizados necesitan estructuras resistentes, mientras que la licuefacción requiere un consumo energético importante y sistemas de aislamiento térmico.

Surge, por tanto, una pregunta más interesante: **¿sería posible almacenar el hidrógeno dentro de un material, en lugar de depender exclusivamente de un depósito lleno de gas comprimido o líquido criogénico?**

Esta posibilidad conduce al almacenamiento de hidrógeno en materiales. Algunos sólidos pueden retener hidrógeno en sus superficies; otros permiten que los átomos de hidrógeno entren en su estructura, y otros forman compuestos químicos que contienen hidrógeno. Estas estrategias no son equivalentes: en cada caso cambian la forma en que se incorpora el hidrógeno, la energía necesaria para liberarlo y las condiciones requeridas para recuperarlo.

El almacenamiento químico resulta especialmente interesante porque permite estudiar el problema desde la propia naturaleza de los enlaces. Un material puede contener una proporción elevada de hidrógeno, pero eso no garantiza que sea útil para un automóvil. Si el hidrógeno queda unido demasiado fuertemente, podría requerirse una temperatura elevada para liberarlo. Si se libera con demasiada facilidad, quizá resulte difícil mantenerlo almacenado de forma estable. Y si el material no puede volver a incorporar el hidrógeno en condiciones razonables, su utilidad como sistema recargable queda limitada.

En consecuencia, el objetivo no consiste simplemente en encontrar el material que contenga más hidrógeno, sino en encontrar un equilibrio entre capacidad, estabilidad, velocidad de reacción, temperatura de operación, consumo energético y seguridad.


## **¿Qué significa almacenar hidrógeno de forma eficiente?**

Para evaluar una propuesta de almacenamiento, es necesario distinguir varios criterios que suelen confundirse.

**La capacidad gravimétrica** indica qué fracción de la masa del material corresponde al hidrógeno almacenado. Si un material contiene una cantidad elevada de hidrógeno, puede parecer atractivo desde esta perspectiva. Sin embargo, la masa de los recipientes, las válvulas, los intercambiadores térmicos y los demás componentes también cuenta cuando se considera el sistema completo del vehículo.

**La capacidad volumétrica** indica cuánto hidrógeno puede almacenarse en un volumen determinado. Un material podría tener una buena capacidad por unidad de masa y, aun así, ocupar demasiado espacio para resultar práctico en un automóvil.

**La temperatura de liberación** determina cuánto debe calentarse el material para recuperar el hidrógeno. Esta variable importa porque el vehículo dispone de recursos energéticos limitados y porque el calor utilizado para liberar el combustible no puede considerarse gratuito.

**La cinética de absorción y liberación** describe la rapidez con la que ocurren los procesos. No basta con que un material pueda almacenar hidrógeno en condiciones ideales: debe hacerlo a una velocidad compatible con el funcionamiento y el repostaje del vehículo.

**La reversibilidad** indica si el material puede liberar el hidrógeno y volver a incorporarlo después de un ciclo de operación. Para un sistema recargable, no es suficiente demostrar una liberación inicial; es necesario estudiar también la recuperación de la capacidad a lo largo de múltiples ciclos.

Finalmente, **la seguridad y el coste** condicionan cualquier aplicación real. Un material atractivo desde el punto de vista químico puede resultar poco conveniente si necesita condiciones de operación difíciles de mantener, si se degrada rápidamente o si su fabricación es demasiado costosa.

Estos criterios revelan que el almacenamiento de hidrógeno es un problema multidimensional. Mejorar una propiedad no garantiza mejorar todas las demás. Precisamente por eso, el estudio de materiales requiere combinar química, física, simulación computacional y evaluación de las condiciones de operación.

## **Del material al vehículo: ¿cuáles son las exigencias reales?**

Para que una tecnología de almacenamiento pueda utilizarse en un vehículo, sus propiedades deben evaluarse a escala de sistema. La capacidad del compuesto aislado no es necesariamente la capacidad del depósito instalado: el conjunto debe incluir el material activo y todos los componentes necesarios para contenerlo, controlar su temperatura, gestionar la presión y permitir la entrada y salida del hidrógeno.

Esta diferencia es importante al comparar los resultados de laboratorio con los objetivos tecnológicos. En Estados Unidos, el Departamento de Energía (DOE) publica objetivos técnicos para sistemas de almacenamiento de hidrógeno destinados al transporte. Estos objetivos permiten evaluar aspectos como la capacidad gravimétrica y volumétrica, la eficiencia, el tiempo de repostaje, la durabilidad y el coste. Por otra parte, el Departamento de Transporte (DOT) y las normas de seguridad aplicables establecen requisitos relacionados con el transporte, los recipientes y la manipulación del hidrógeno.

No se trata de dos maneras de expresar el mismo criterio. **Los objetivos técnicos del DOE orientan el desarrollo de tecnologías; los requisitos de seguridad y transporte establecen condiciones que deben cumplirse en los ámbitos correspondientes.** Un material no demuestra que sea apto para un vehículo simplemente porque alcance una capacidad elevada en un experimento, ni un resultado de laboratorio equivale a la certificación de un depósito comercial.

Por ello, cualquier propuesta debe responder dos preguntas diferentes: ¿puede el material almacenar y liberar hidrógeno con propiedades interesantes?, y ¿podría integrarse en un sistema vehicular que satisfaga las exigencias de rendimiento, seguridad y coste?

La primera pregunta puede investigarse mediante experimentos y cálculos atomísticos. La segunda exige considerar también el diseño del sistema completo y las condiciones reales de operación.

## **La pregunta que orienta este trabajo**

A partir de estas dificultades, el problema deja de ser simplemente cómo introducir más hidrógeno en un depósito. Se convierte en una pregunta sobre la materia: ¿qué características químicas y estructurales permiten que un sólido incorpore hidrógeno y lo libere de manera controlada, sin exigir condiciones incompatibles con una aplicación práctica?

Para explorar esta cuestión, estudiaremos el borohidruro de litio, cuya fórmula química es \(LiBH_4\), y la posible influencia del níquel, \(Ni\), sobre los procesos relacionados con el hidrógeno. El interés de esta combinación no reside únicamente en la presencia de hidrógeno dentro de un compuesto químico, sino en comprender cómo la composición, la estructura cristalina y las interacciones con superficies metálicas pueden modificar el comportamiento del material.

La investigación avanzará desde la evidencia experimental publicada hasta la construcción de modelos atomísticos y el análisis de sus propiedades mediante cálculos de estructura electrónica. Cada etapa deberá responder una pregunta concreta y, al mismo tiempo, establecer los límites de lo que realmente puede concluirse.

El propósito no será demostrar de antemano que $L i B H_{4}$ combinado con $Ni$ constituye una solución definitiva para los vehículos. Será investigar qué hace interesante a este sistema, qué mecanismos podrían explicar su comportamiento y hasta qué punto la evidencia disponible permite considerarlo una alternativa prometedora.

La primera incógnita, entonces, es inevitable: **¿por qué estudiar precisamente el borohidruro de litio y el níquel, cuando existen tantas maneras diferentes de almacenar hidrógeno?** Para responderla, primero debemos comparar las principales estrategias y examinar qué ha encontrado la investigación científica sobre esta combinación.

# **2. Investigación previa y selección de LiBH₄–Ni**


## **¿Qué alternativas existen para almacenar hidrógeno?**

La introducción nos dejó una pregunta abierta: ¿qué tipo de material permitiría almacenar hidrógeno y recuperarlo cuando el vehículo lo necesite? Antes de elegir un compuesto específico, conviene examinar las principales estrategias que se han investigado. Cada una resuelve una parte del problema, pero también introduce limitaciones que deben considerarse.

Una primera posibilidad es almacenar el hidrógeno como gas comprimido. Esta tecnología evita depender de una reacción química para liberar el combustible, pero exige recipientes capaces de soportar altas presiones. Otra alternativa consiste en almacenar hidrógeno líquido a temperaturas criogénicas. Aunque permite aumentar considerablemente la cantidad de hidrógeno por unidad de volumen respecto al gas a presión atmosférica, requiere energía para la licuefacción y medidas para controlar las pérdidas térmicas.

Una tercera estrategia consiste en utilizar materiales porosos, como determinados sólidos con una gran superficie interna, sobre los cuales pueden adsorberse moléculas de hidrógeno. En este caso, la cantidad almacenada depende de factores como la superficie disponible, la temperatura y la presión. El desafío es conseguir una capacidad suficiente sin que el sistema necesite condiciones de operación poco prácticas.

Finalmente, existen materiales que incorporan hidrógeno mediante interacciones más intensas o reacciones químicas. Algunos metales y aleaciones pueden formar hidruros; otros compuestos, conocidos como hidruros complejos, contienen hidrógeno unido a elementos como el boro. Estas familias son interesantes porque pueden almacenar cantidades importantes de hidrógeno por unidad de masa. Sin embargo, liberarlo y volver a incorporarlo puede exigir temperaturas elevadas, presiones considerables o tiempos de reacción demasiado largos.

No existe, por tanto, una estrategia universalmente superior. La elección depende de las condiciones de funcionamiento y de la importancia relativa de la capacidad, la seguridad, el coste y la facilidad de recarga.

Para este trabajo nos interesa especialmente el almacenamiento químico en sólidos. Su estudio permite plantear una pregunta más específica: **¿podemos modificar la estructura y el entorno químico de un compuesto para que almacene hidrógeno y lo libere en condiciones más favorables?**

## **El borohidruro de litio: una capacidad atractiva con un problema importante**

Entre los materiales investigados para almacenar hidrógeno se encuentra el borohidruro de litio, cuya fórmula es $LiBH_4$. Este compuesto pertenece a la familia de los hidruros complejos y contiene el anión borohidruro, $BH_4^-$, asociado con cationes de litio, $Li^+$.

Su principal atractivo es la elevada proporción de hidrógeno que contiene. De acuerdo con el artículo de Zhang y colaboradores, el $LiBH_4$ posee una densidad gravimétrica teórica de hidrógeno de aproximadamente $18.5$% en masa. Es decir, una fracción importante de la masa del compuesto corresponde a hidrógeno. Esta propiedad lo convierte en un candidato interesante para sistemas en los que cada kilogramo cuenta.

Sin embargo, una elevada capacidad teórica no significa que todo ese hidrógeno pueda extraerse de manera útil, rápida y reversible. El $LiBH_4$ presenta una estabilidad termodinámica elevada y barreras cinéticas que dificultan su funcionamiento como sistema recargable en condiciones moderadas. En otras palabras, el material puede contener mucho hidrógeno, pero recuperarlo y volver a almacenarlo constituye un problema químico considerable.

Aquí aparece una distinción fundamental. La **capacidad de almacenamiento** responde a cuánto hidrógeno puede contener un material; la **cinética** describe con qué rapidez lo incorpora o lo libera; y la **termodinámica** permite estudiar qué estados son energéticamente favorables y bajo qué condiciones pueden establecerse. Mejorar una de estas propiedades no garantiza mejorar las demás.

Por ejemplo, si los enlaces y las estructuras que mantienen al hidrógeno dentro del material son demasiado estables, puede ser necesario aportar demasiado calor para liberarlo. Pero debilitar excesivamente esas interacciones tampoco sería necesariamente deseable, porque el almacenamiento debe ser suficientemente estable y controlable.

Por eso, el interés científico no se limita a encontrar un compuesto con una gran cantidad de hidrógeno. También consiste en comprender qué cambios en su entorno químico podrían facilitar las transformaciones necesarias para almacenarlo y recuperarlo.

## **¿Por qué introducir níquel?**

Una estrategia para modificar el comportamiento de un material consiste en incorporar un segundo componente que influya en sus reacciones. En este contexto aparece el níquel, $Ni$, un metal de transición cuyas propiedades electrónicas y superficiales pueden favorecer determinados procesos catalíticos.

En sistemas relacionados con el hidrógeno, las superficies metálicas pueden facilitar la ruptura de enlaces de las moléculas de hidrógeno, \(H_2\), y modificar las interacciones entre los átomos de hidrógeno y el material que los recibe. Esto resulta relevante porque la incorporación de hidrógeno molecular no consiste simplemente en acercar una molécula a un sólido: en ciertos mecanismos, es necesario romper el enlace \(H-H\), reorganizar los enlaces del material y formar nuevas interacciones químicas.

No obstante, debemos evitar una conclusión prematura. Que el níquel pueda actuar como catalizador en determinados procesos no demuestra, por sí solo, que cualquier mezcla de \(Ni\) y \(LiBH_4\) vaya a almacenar más hidrógeno o a liberarlo a menor temperatura. El efecto depende de la composición, el tamaño de las partículas, la distribución de los componentes y la forma en que entran en contacto.

La hipótesis que merece investigarse es más concreta: **si el níquel se coloca en el entorno adecuado del borohidruro de litio, ¿puede modificar las interacciones responsables de las dificultades de absorción y liberación del hidrógeno?**

Para responderla, necesitamos una investigación que vaya más allá de las propiedades conocidas de cada componente por separado.

## **El artículo de referencia: una arquitectura de tres capas**

Una investigación publicada en *Nature Communications* en marzo de $2026$ ofrece un punto de partida especialmente relevante. El artículo, titulado "*Reversible H₂ storage at moderate temperature by a trilayered lithium borohydride nanocomposite*", fue elaborado por Zhang y colaboradores. Los autores estudiaron un nanocompuesto diseñado para mejorar el almacenamiento reversible de hidrógeno en $LiBH_4$. [Leer el artículo original](https://doi.org/10.1038/s41467-026-69059-y).

La propuesta no consiste simplemente en mezclar polvo de níquel con borohidruro de litio. El material experimental presenta una arquitectura organizada en tres componentes: un soporte de grafeno, una capa de nanoclústeres de níquel y nanopartículas de $LiBH_4$. En términos esquemáticos, la disposición puede representarse así:

$$\text{Grafeno}\;|\;\text{Ni}\;|\;LiBH_4$$

Cada componente cumple una función propuesta dentro del diseño. El grafeno sirve como soporte para distribuir los nanoclústeres metálicos y favorecer la formación de nanopartículas del hidruro. El níquel proporciona una superficie de contacto entre el soporte y el \(LiBH_4\), mientras que el borohidruro de litio constituye la fase principal de almacenamiento químico del hidrógeno.

La composición que los autores identificaron como favorable contenía aproximadamente $70$% en masa de $LiBH_4$, $27$% de grafeno y $3$% de níquel. Estas proporciones corresponden al nanocompuesto experimental estudiado, no a una mezcla universalmente óptima ni a un sistema vehicular completo.

¿Por qué importa tanto la disposición espacial? Porque dos materiales con la misma composición global pueden comportarse de manera diferente si sus componentes están organizados de otra forma. El contacto entre las fases determina qué átomos pueden interactuar, qué superficies quedan expuestas y qué transformaciones químicas resultan más accesibles.

En el sistema estudiado, los autores proponen que la presencia de níquel en contacto directo con el $LiBH_4$, mientras reduce el contacto directo entre este último y el grafeno, favorece las transformaciones relacionadas con la incorporación y liberación del hidrógeno. La arquitectura no es, por tanto, un detalle decorativo: forma parte de la hipótesis química que el trabajo intenta demostrar.

## **¿Qué encontraron los investigadores?**

Los resultados experimentales son prometedores porque muestran mejoras en varias propiedades importantes del sistema.

En primer lugar, los autores reportaron una temperatura de inicio de liberación de hidrógeno de aproximadamente \(100\,^\circ\mathrm{C}\) para el nanocompuesto seleccionado. La temperatura del máximo de liberación durante el calentamiento fue de aproximadamente \(330\,^\circ\mathrm{C}\), frente a unos \(435\,^\circ\mathrm{C}\) para el \(LiBH_4\) puro en las condiciones de comparación descritas en el artículo.

Estos valores representan dos medidas diferentes. La temperatura de inicio señala cuándo comienza a detectarse el proceso bajo el protocolo experimental utilizado; la temperatura del máximo corresponde al punto de mayor intensidad de liberación durante el ensayo de calentamiento. Ninguna de las dos debe interpretarse automáticamente como la temperatura necesaria para operar un depósito vehicular completo.

En segundo lugar, el nanocompuesto mostró una capacidad reversible reportada de hasta aproximadamente $12.3$% en masa de hidrógeno respecto a la masa de $LiBH_4$, bajo las condiciones experimentales descritas por los autores, que incluyen una presión de hidrógeno de $100$ bar. Esta referencia de masa es esencial: **no significa que el sistema completo contenga $12.3$ de hidrógeno respecto a su propia masa, ni que un depósito de automóvil alcance esa capacidad gravimétrica útil**.

El artículo también reporta una temperatura de inicio de absorción cercana a $70$°. Además, los investigadores observaron que el material podía reincorporar una cantidad importante de hidrógeno a temperaturas inferiores a las necesarias para completar el proceso en condiciones más exigentes. En uno de los ensayos, la absorción alcanzó aproximadamente $8.1$% en masa de hidrógeno a $150$, bajo las condiciones y el tiempo de permanencia indicados por los autores.

Estos resultados sugieren que el sistema mejora tanto la liberación como la reincorporación del hidrógeno en comparación con el $LiBH_4$ sin modificar. Sin embargo, las cifras deben leerse junto con las condiciones de presión, temperatura, duración y normalización de masa de cada ensayo. No todas describen el mismo experimento ni representan una única capacidad operativa.

La importancia de estos hallazgos no reside exclusivamente en alcanzar una cifra elevada. Lo relevante es que la organización del material parece modificar las condiciones en las que se producen procesos que, en el compuesto original, resultan difíciles de controlar.

## **¿Qué mecanismo podría explicar la mejora?**

Los experimentos permiten comparar el comportamiento del material antes y después de introducir los componentes adicionales. Pero observar una mejora no explica por sí mismo su causa. Para avanzar desde la correlación experimental hacia una explicación química, los autores combinaron la caracterización del material con cálculos basados en la teoría del funcional de la densidad, o DFT.

En sus modelos teóricos, los investigadores estudiaron un agregado de boro formado por $72$ átomos, denominado $B_{72}$, y compararon su comportamiento en diferentes entornos, incluyendo superficies de carbono y níquel. Estos modelos permitieron investigar cómo cambia la estructura del boro al interactuar con el níquel y cómo esas modificaciones afectan las energías de interacción con el hidrógeno.

Los cálculos indican que el contacto con la superficie de níquel puede provocar una reorganización importante del agregado de boro, incluida la ruptura de algunos enlaces $B-B$. Según la interpretación de los autores, esta reorganización crea entornos adicionales en los que el hidrógeno puede adsorberse y modifica la energía asociada con su interacción con los átomos de boro.

El artículo también propone que el níquel facilita procesos relacionados con la ruptura del enlace $H-H$, la incorporación del hidrógeno y su migración por el sistema. En conjunto, estos efectos ofrecen una explicación posible de la mejora observada experimentalmente.

Pero hay una limitación que debemos conservar desde el principio: el modelo $B_{72}$ empleado en los cálculos no equivale a una simulación atomística completa del nanocompuesto experimental. No contiene de manera explícita todos los componentes, las nanopartículas reales y las transformaciones químicas del sistema completo de $LiBH_4$. Los cálculos permiten investigar mecanismos locales y formular explicaciones fundamentadas, pero no reproducen por sí solos todo el ciclo de almacenamiento.

Por tanto, conviene distinguir tres niveles de conocimiento:

- **Resultado experimental:** el nanocompuesto presenta mejoras medidas en las condiciones específicas descritas en el artículo.
- **Resultado computacional:** los modelos de DFT muestran cómo determinadas interacciones con el níquel pueden modificar la estructura del boro y las energías de adsorción del hidrógeno.
- **Interpretación mecanística:** los autores relacionan ambos conjuntos de resultados y proponen que estos cambios ayudan a explicar la mejora en la reversibilidad.

Esta distinción será importante en nuestro propio trabajo. No debemos presentar como observación directa aquello que es una inferencia basada en un modelo, ni asumir que un cálculo sobre una estructura simplificada demuestra automáticamente el comportamiento de un material real.

## **¿Es esta solución adecuada para un automóvil?**


Los resultados del artículo justifican estudiar el sistema, pero todavía no permiten afirmar que esté listo para utilizarse en un vehículo.

Para entender por qué, podemos compararlos con los objetivos técnicos que publica el Departamento de Energía de Estados Unidos para sistemas de almacenamiento de hidrógeno destinados a vehículos ligeros. Estos objetivos consideran la capacidad gravimétrica y volumétrica del sistema completo, el coste, la durabilidad, la eficiencia y el tiempo de repostaje, entre otros parámetros. [Consultar los objetivos técnicos del DOE](https://www.energy.gov/cmei/fuels/doe-technical-targets-onboard-hydrogen-storage-light-duty-vehicles).

La distinción entre material y sistema vuelve a ser decisiva. En un vehículo, el almacenamiento debe incluir no solo el compuesto activo, sino también el recipiente, las válvulas, las tuberías, los elementos de control y los componentes necesarios para gestionar la temperatura y la presión. La masa y el volumen de estos elementos reducen la capacidad útil respecto a la calculada exclusivamente para el material.

Además, el uso de una presión de $100$ bar en los experimentos de absorción plantea una pregunta de ingeniería: ¿qué infraestructura y qué condiciones de operación serían necesarias para realizar la recarga de forma práctica? # **3. Fundamentos químicos y cuánticos.**

## **De la composición química a las interacciones atómicas**

En la sección anterior identificamos un problema: el borohidruro de litio, $LiBH_4$, contiene una proporción elevada de hidrógeno, pero su estabilidad dificulta liberarlo y reincorporarlo en condiciones moderadas. También encontramos que la introducción de níquel puede modificar el comportamiento del sistema. Sin embargo, todavía queda una pregunta fundamental: **¿qué ocurre realmente entre los átomos para que un metal pueda cambiar la manera en que otro material interactúa con el hidrógeno?**

Para responderla, necesitamos pasar de las propiedades macroscópicas —como la temperatura de liberación o la capacidad de almacenamiento— a la escala atómica. Allí, las propiedades de un material dependen de su composición, la disposición de sus átomos y, especialmente, de la distribución de sus electrones.

Esta transición es esencial para la química computacional. No basta con conocer qué elementos forman un compuesto; debemos comprender cómo se enlazan, qué estados electrónicos pueden ocupar y cómo cambia su distribución electrónica cuando interactúan con otros átomos.

## **¿Qué hace especial al borohidruro de litio?**

La fórmula química del borohidruro de litio es $LiBH_4$. En él encontramos tres elementos: litio, boro e hidrógeno. Aunque su fórmula contiene cuatro átomos de hidrógeno por unidad fórmula, estos no se encuentran simplemente atrapados en un espacio vacío. Forman parte de una estructura química organizada.

Para comprenderla, podemos separar conceptualmente el compuesto en dos unidades:


$$LiBH_4 \approx Li^+ + BH_4^-$$


Esta representación destaca su carácter iónico: el litio se describe como un catión, $Li^+$, y el grupo borohidruro como un anión, $BH_4^-$. En el sólido, la atracción electrostática entre estas especies contribuye a mantener la estructura cristalina.

Sin embargo, hay una distinción importante. Mientras la interacción entre $Li^+$ y $BH_4^-$ se describe principalmente en términos iónicos, los enlaces entre el boro y los cuatro átomos de hidrógeno dentro del anión son enlaces covalentes polares. Los electrones se comparten entre el boro y el hidrógeno, aunque su distribución no es idéntica alrededor de ambos elementos.

El grupo $BH_4^-$ presenta una geometría aproximadamente tetraédrica: el átomo de boro se encuentra rodeado por cuatro átomos de hidrógeno. Esta disposición no es arbitraria; está relacionada con la organización de los orbitales y la distribución electrónica que estabiliza la especie.

Podemos imaginar el material como una estructura en la que los grupos $BH_4^-$ mantienen unidos sus átomos de hidrógeno mediante enlaces covalentes, mientras que los cationes de litio contribuyen a organizar y estabilizar el sólido mediante interacciones electrostáticas.

Esta descripción permite entender por qué almacenar hidrógeno en un compuesto químico es diferente de almacenarlo como gas comprimido. En el gas molecular, los átomos de hidrógeno están unidos en moléculas $H_2$. En el borohidruro, el hidrógeno forma parte de grupos químicos cuya estructura debe transformarse para que el hidrógeno pueda liberarse y recuperarse.

**La primera dificultad, entonces, no consiste únicamente en atraer hidrógeno hacia el material, sino en modificar las interacciones químicas que determinan dónde se encuentra y cómo puede desplazarse.**


## **Los enlaces químicos desde la perspectiva cuántica**

Para comprender estas interacciones debemos considerar el comportamiento de los electrones. En la descripción cuántica, los electrones no recorren órbitas clásicas perfectamente definidas alrededor del núcleo. Se describen mediante estados cuánticos y orbitales, que permiten representar la distribución espacial de la probabilidad de encontrarlos.

Los orbitales atómicos se clasifican habitualmente como $s$, $p$, $d$ y $f$, entre otros. Cada tipo presenta características espaciales y energéticas diferentes. La distribución de los electrones entre estos estados contribuye a determinar las propiedades químicas de cada elemento.

En el caso del litio, el boro y el hidrógeno, los electrones de valencia participan en la formación de los enlaces que mantienen unido el anión borohidruro y en las interacciones que estabilizan el sólido. La distribución electrónica del conjunto no puede entenderse únicamente examinando cada átomo por separado: cuando los átomos se aproximan, sus estados electrónicos interactúan y dan lugar a nuevos estados asociados al sistema enlazado.

La formación de un enlace puede interpretarse como una reorganización electrónica que, bajo las condiciones adecuadas, hace energéticamente favorable la configuración enlazada respecto a determinados estados separados. Romperlo requiere modificar esa configuración y, generalmente, aportar energía o encontrar una ruta de reacción que reduzca la barrera que debe superarse.

Esto introduce dos conceptos que aparecerán repetidamente en nuestro trabajo: **la energía de los estados electrónicos y la barrera energética de una transformación química**.

No son lo mismo. Una reacción puede conducir a un estado final energéticamente favorable y, aun así, ocurrir lentamente si existe una barrera elevada entre el estado inicial y el final. Esta distinción ayudará a comprender por qué un material puede ser capaz de almacenar hidrógeno, pero necesitar condiciones exigentes para liberarlo.

## **¿Por qué importa que el níquel sea un metal de transición?**


El níquel, cuyo símbolo es $Ni$, es un metal de transición. Una característica relevante de los metales de transición es que sus estados electrónicos $d$ desempeñan un papel importante en numerosas de sus propiedades químicas, especialmente en las interacciones con moléculas y átomos adsorbidos sobre sus superficies.

En el átomo aislado, la configuración electrónica fundamental del níquel se representa aproximadamente como:

$$[Ar]\,3d^8\,4s^2$$

Esta notación describe la ocupación electrónica del átomo aislado. No debe interpretarse como una descripción completa de los electrones del níquel metálico: en un sólido, los estados de los átomos vecinos interactúan y forman estados electrónicos extendidos, cuyas energías y ocupaciones dependen de la estructura cristalina y del entorno.

En una superficie de níquel, los estados electrónicos disponibles pueden interactuar con los orbitales de una molécula de hidrógeno o con los orbitales de átomos de boro e hidrógeno presentes cerca de ella. Estas interacciones pueden modificar la distribución electrónica de las especies involucradas y cambiar la estabilidad relativa de diferentes configuraciones atómicas.

Por eso, el níquel resulta interesante como posible modificador catalítico. Sus propiedades electrónicas permiten investigar si determinadas interacciones con el hidrógeno o con el boro facilitan procesos que serían menos favorables en ausencia del metal.

Sin embargo, debemos evitar una simplificación frecuente: **tener orbitales $d$ no garantiza automáticamente que un metal sea un buen catalizador para cualquier reacción**. La actividad depende de la estructura de la superficie, la coordinación de los átomos, los estados electrónicos disponibles, las especies adsorbidas y las energías de los diferentes estados del proceso.

En nuestro caso, no basta con afirmar que el níquel facilita la liberación de hidrógeno porque es un metal de transición. Necesitamos estudiar cómo se relaciona con el boro y el hidrógeno y comprobar si los cambios calculados son compatibles con el mecanismo propuesto.

## **Adsorción, absorción y almacenamiento químico: tres conceptos diferentes**


Antes de estudiar las interacciones entre níquel e hidrógeno, conviene distinguir tres procesos que pueden parecer similares, pero que describen fenómenos diferentes.

La **adsorción** ocurre cuando una especie se une o interactúa con una superficie. Por ejemplo, una molécula de hidrógeno puede acercarse a una superficie de níquel e interactuar con sus estados electrónicos. Dependiendo de la superficie y de las condiciones, la molécula puede permanecer intacta o experimentar una transformación, como la ruptura de su enlace.

La **absorción** ocurre cuando una especie penetra en el interior de otra fase. En determinados materiales, los átomos de hidrógeno pueden incorporarse a posiciones dentro de la estructura, en lugar de permanecer únicamente en la superficie.

El **almacenamiento químico**, por su parte, implica que el hidrógeno forma parte de una especie o fase química y que su incorporación o liberación requiere transformaciones químicas. En el $LiBH_4$, el hidrógeno forma parte de los grupos borohidruro; su almacenamiento no se reduce a la adsorción de moléculas intactas sobre una superficie.

Estas distinciones son esenciales para interpretar los cálculos. Si un modelo demuestra que el hidrógeno se adsorbe favorablemente sobre una superficie de níquel, eso no demuestra por sí solo que el sistema completo de $LiBH_4$ pueda almacenar una cantidad determinada de hidrógeno de manera reversible.

Lo que sí puede aportar es información sobre una etapa posible del mecanismo: cómo interactúa el hidrógeno con una superficie y qué configuraciones podrían facilitar procesos posteriores.

## **La energía de adsorción: ¿qué tan favorable es una interacción?**

Una de las primeras propiedades que podemos calcular es la energía de adsorción. Esta magnitud permite comparar la energía de una superficie con hidrógeno adsorbido frente a las energías de la superficie limpia y de la especie de hidrógeno de referencia.

Para estudiar la adsorción de un átomo de hidrógeno sobre una superficie de níquel, utilizaremos la siguiente convención:

$$E_{\mathrm{ads}}=E_{\mathrm{Ni+H}}-E_{\mathrm{Ni}}-\frac{1}{2}E_{H_2}$$

Aquí, $E_{\mathrm{Ni+H}}$ representa la energía total del modelo de níquel con el átomo de hidrógeno adsorbido; $E_{\mathrm{Ni}}$, la energía de la superficie limpia; y $E_{H_2}$, la energía de una molécula aislada de hidrógeno. El factor $\frac{1}{2}$ permite utilizar como referencia un átomo de hidrógeno procedente de una molécula $H_2$.

Con esta convención, un valor negativo de $E_{\mathrm{ads}}$ indica que la configuración adsorbida tiene menor energía que la superficie limpia más la cantidad correspondiente de hidrógeno molecular, dentro de las condiciones y aproximaciones del cálculo. Un valor positivo indica que la configuración adsorbida es energéticamente desfavorable respecto a esa referencia.

No obstante, hay que interpretar la magnitud con cuidado. Una energía de adsorción muy negativa no significa necesariamente que el material sea mejor para el almacenamiento reversible. Si el hidrógeno queda demasiado estabilizado en una configuración, podría ser difícil liberarlo. Además, la energía de adsorción no es lo mismo que una barrera de activación ni equivale directamente a una temperatura de liberación.

Por tanto, la energía de adsorción es una primera herramienta para investigar la interacción entre el hidrógeno y una superficie, pero no constituye por sí sola una medida completa de la calidad de un sistema de almacenamiento.

## **¿Cómo podemos observar estas interacciones mediante cálculos?**

Las interacciones electrónicas no se observan directamente como si fueran objetos macroscópicos. Para estudiarlas computacionalmente necesitamos resolver un modelo físico y analizar magnitudes que permitan caracterizar la estructura y la energía del sistema.

La teoría del funcional de la densidad, conocida como DFT, es una herramienta de la mecánica cuántica que permite calcular aproximaciones a la energía electrónica y a la distribución de electrones de un sistema formado por átomos. A partir de estas cantidades, es posible comparar configuraciones, optimizar geometrías y estudiar propiedades relacionadas con la formación de enlaces.

En nuestro proyecto, algunos de los análisis más relevantes serán:

- **Geometría optimizada:** permite examinar las posiciones atómicas y las distancias entre los átomos después de minimizar la energía del modelo.
- **Energía de adsorción:** permite comparar la estabilidad de distintas configuraciones de hidrógeno sobre una superficie.
- **Densidad electrónica:** representa cómo se distribuye la carga electrónica en el espacio y puede ayudar a estudiar los cambios asociados con una interacción.
- **Densidad de estados electrónicos (DOS y PDOS):** muestra cómo se distribuyen los estados electrónicos según su energía. La PDOS permite analizar las contribuciones asociadas a determinados átomos u orbitales.
- **Barreras de migración o reacción:** permiten estimar la energía que debe superarse para pasar de una configuración inicial a otra final mediante una ruta determinada.

Cada herramienta responde una pregunta diferente. Una geometría optimizada permite estudiar distancias y coordinación, pero no demuestra por sí sola que una reacción sea rápida. Una energía de adsorción compara estados iniciales y finales, pero no proporciona automáticamente la barrera entre ellos. Una densidad electrónica puede mostrar cambios en la distribución de carga, pero su interpretación debe relacionarse con otras evidencias.

Por eso, la investigación no se apoyará en una única cifra. Buscaremos relacionar la geometría, las energías y las propiedades electrónicas para construir una explicación coherente y reconocer qué aspectos siguen siendo inciertos.
